## 1. My lane as an ML task

### Lane: Refresh / Content Opportunity Scoring

My chosen lane is **Refresh / Content Opportunity Scoring**.

The ML task type is **scoring**.

The goal is to assign each content page an opportunity score that estimates how strongly that page should be considered for a content action such as:

- Refresh
- Expand
- Protect
- Prune
- Monitor

The unit of analysis is **one content page**.

Instead of using a single fixed rule such as "refresh every page with low traffic", the model can combine multiple signals such as search demand, competition, content length, CTR, average position, engagement, scroll behaviour, AI traffic, and trend information.

The output of the model would be a **ranked review queue** of content pages, so that the highest-priority opportunities can be reviewed first.

## 3. Success metric

The main success metric for this scoring task is **Precision@K**.

K represents the number of pages the content team can realistically review in a given period. For example, if the team can review 100 pages, Precision@100 measures the proportion of those top 100 scored pages that later show a meaningful positive outcome after the content action.

This metric fits the task because the output is a ranked list and the content team has limited capacity.

For the future supervised version of this task, a good result would mean that a high proportion of the pages near the top of the ranked list actually benefit from a refresh.

A secondary business measure could be the change in sessions or engagement after a refresh, but **Precision@K** is the primary metric for evaluating the ranking.

```python
# Check the number of content pages available for ranking

print("Number of content pages:", len(df))
print("Example review capacity (K): 100")
print("Primary success metric: Precision@100")
```

## 5. Why ML beats a fixed rule here

## 5. Why ML beats a fixed rule here

A fixed rule might say:

> "Refresh every page with a low CTR."

The problem is that content opportunity depends on multiple signals at the same time.

For example, a page can have a low CTR but also have high search volume, a strong average position, increasing demand, and good engagement. Another page may have low traffic simply because there is very little search demand.

The dataset contains multiple signals including search volume, competition, content length, CTR, average position, engagement rate, scroll rate, AI traffic, freshness, and recent trends.

A single if-statement would require manually choosing thresholds for these variables and would not easily capture how they interact.

ML is useful here because a future model could learn relationships between these signals and a measured post-refresh outcome from historical data.

The model would therefore provide **decision support** by producing a ranked opportunity list. Human review would still determine whether a page should actually be refreshed, expanded, protected, pruned, or monitored.

The purpose of ML is not to automatically make the content decision, but to prioritize which pages deserve attention first.
